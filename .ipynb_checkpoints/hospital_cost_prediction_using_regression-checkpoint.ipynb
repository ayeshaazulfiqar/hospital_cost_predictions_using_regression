## **Linear Regression**

```python
from sklearn.linear_model import LinearRegression
#independent and dependent variables:
y = data['annual_medical_cost']
X = data[['monthly_premium','annual_premium','total_claims_paid','avg_claim_amount','risk_score','chronic_count']]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
#using test train method
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42)
```

```python
#standerdizing the dataset
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train= scaler.fit_transform(X_train)  #fittransform is used on training data to learn parameters and to transform it
```

```python
X_test= scaler.transform(X_test) #transform is used on test data to apply previously learned transformations
```

```python
#cross validation
from sklearn.model_selection import cross_val_score
regression=LinearRegression()
fit=regression.fit(X_train,y_train)
joblib.dump(regression, "linear_regression_model.joblib")
```

```python
#mean squared error 
from sklearn.metrics import (explained_variance_score,mean_absolute_error,mean_squared_error, median_absolute_error,r2_score)
mse=cross_val_score(regression, X_train,y_train,scoring= 'neg_mean_squared_error',cv=5) #here cv=5 means i am doing cross validation five times means my model is training 5 times with different different data
```

```python
np.mean(mse) #it should be less and less bcz it is the difference between predicted and truth value
```

```python
#prediction
reg_pred=regression.predict(X_test)
```

```python
sns.displot(reg_pred-y_test,kind='kde') #i am plotting the subtracted value of predicted variables and test values
```

The variance here is quite low

```python
score=r2_score(reg_pred,y_test)
```

```python
score
```

## **Ridge Regression**

1: we use ridge regression to handle the overfitting issues
2: we do hyper paramter tuning for alpha or lambda amd we use method of gridsearchCV

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
```

```python
ridge_regressor=Ridge()
```

```python
#now by default alpha is 1.0 but we want alpha according to our problem
parameter={'alpha':[0.001, 0.01, 0.1, 1, 10]}
ridgecv = GridSearchCV(ridge_regressor,parameter,scoring='neg_mean_squared_error',cv=5)
ridgecv.fit(X_train,y_train)
joblib.dump(ridgecv.best_estimator_, "ridge_regression_model.joblib")
```

```python
print(ridgecv.best_params_) #it will select the best alpha value according to our problem
```

```python
print(ridgecv.best_score_) # a very slight difference in the error in comparison to linear regression mse
```

```python
ridge_pred=ridgecv.predict(X_test)
ridge_pred
```

```python
sns.displot(ridge_pred-y_test,kind='kde')
```

The variance here is quite low as well which means my model is performing well

```python
score=r2_score(ridge_pred,y_test)
```

```python
score
```

## **Lasso Regression**

This is same as ridge regression only difference is in the parameters of the function lasso, here we only take absolute value of lasso instead of squared value that we used in ridge regression

```python
from sklearn.linear_model import Lasso
```

```python
lasso=Lasso()
```

```python
#now by default alpha is 1.0 but we want alpha according to our problem
parameters={'alpha':[0.001, 0.01, 0.1, 1, 10]}
lassocv = GridSearchCV(lasso,parameters,scoring='neg_mean_squared_error',cv=5)
lassocv.fit(X_train,y_train)
joblib.dump(lassocv.best_estimator_, "lasso_regression_model.joblib")
```

```python
print(lassocv.best_params_)
print(lassocv.best_score_)
```

```python
lasso_pred=lassocv.predict(X_test)
lasso_pred
sns.displot(lasso_pred-y_test,kind='kde')
```

```python
score=r2_score(lasso_pred,y_test)
score
```

## **Elastic Net Regression**

```python
from sklearn.linear_model import ElasticNet
```

```python
elasticnet=ElasticNet()
```

```python
#now by default alpha is 1.0 but we want alpha according to our problem
from sklearn.pipeline import Pipeline
pipeline = Pipeline([('scaler', StandardScaler()),('elasticnet', ElasticNet(max_iter=5000, tol=1e-2))])
parameters = {'elasticnet__alpha': [0.001, 0.01, 0.1, 1, 10],'elasticnet__l1_ratio': [0.2, 0.5, 0.8]} #l1_ratio controls the l1 value of ridge and lasso, if l1_ratio=0 then it is pure ridge and if l1_ratio=1 then it is pure lasso, 
elasticcv = GridSearchCV(pipeline,parameters,scoring='neg_mean_squared_error',cv=3,n_jobs=-1)
elasticcv.fit(X_train, y_train)
joblib.dump(pipeline, "elasticnet_model.joblib")
```

## **Gradient Boost Regression**

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
y = data['annual_medical_cost']
X = data[['monthly_premium','annual_premium','total_claims_paid','avg_claim_amount','risk_score','chronic_count']]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42)
regr = GradientBoostingRegressor()
regr.fit(X_train, y_train)
joblib.dump(regr, "gradient_boosting_model.joblib")
gradientboost_pred = regr.predict(X_test)
```

```python
score=r2_score(gradientboost_pred,y_test)
score
```

```python
sns.displot(gradientboost_pred-y_test,kind='kde')
```

```python
# Plot actual vs predicted
plt.figure(figsize=(12, 6))
plt.scatter(range(len(y_test)), y_test, color='black', label='Actual')
plt.plot(gradientboost_pred, color='blue', linewidth=3, label='Predicted')
plt.xlabel("Sample Index")
plt.ylabel("annual_medical_costs")
plt.legend()
plt.show()
# Print evaluation results
print(f"R2 score: {r2_score(y_test, gradientboost_pred):.6f}")
print("Gradient Boosting does not provide coefficients or intercept.")
```

```python
def split_train_test(data, feature, train_index=0.7):
    train, test = train_test_split(data, test_size=1-train_index)

    if isinstance(feature, list):
        x_train = train[feature].to_numpy()
        y_train = train['annual_medical_cost'].to_numpy()
        x_test = test[feature].to_numpy()
        y_test = test['annual_medical_cost'].to_numpy()
    else:
        x_train = train[[feature]].to_numpy()
        y_train = train['annual_medical_cost'].to_numpy()
        x_test = test[[feature]].to_numpy()
        y_test = test['annual_medical_cost'].to_numpy()

    return x_train, y_train, x_test, y_test
```

```python
plt.figure(figsize=(15, 7))
plot_count = 1

for feature in ['monthly_premium','annual_premium','total_claims_paid','avg_claim_amount','risk_score','chronic_count']:
    # Prepare data
    data_tr = data[['annual_medical_cost', feature]]
    data_tr = data_tr[(data_tr.T != 0).all()]
    x_train, y_train, x_test, y_test = split_train_test(data_tr, feature)
    # Create and train Gradient Boosting Regressor
    regr = GradientBoostingRegressor()
    regr.fit(x_train, y_train)
    y_pred = regr.predict(x_test)
    # Plot actual vs predicted
    plt.subplot(2, 3, plot_count)
    plt.scatter(x_test, y_test, color='black')
    plt.plot(x_test, y_pred, color='blue', linewidth=3)
    plt.xlabel(feature.replace('_', ' ').title())
    plt.ylabel('annual_medical_cost')

    # Print R2 score
    print(f"{feature} R2 score: {r2_score(y_test, y_pred):.6f}")

    plot_count += 1

plt.tight_layout()
plt.show()
```

the blur line shows the graph of predicted annual_medical_cost with X_test
